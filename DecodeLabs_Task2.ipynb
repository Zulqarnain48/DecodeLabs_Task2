## Exploratory Data Analysis (EDA)

### 1. Load Data and Initial Inspection

This first step involves loading the Excel dataset into a pandas DataFrame, which is a tabular data structure commonly used for data manipulation and analysis in Python. After loading, we display the first few rows to get a quick overview of the data's structure and content.

```python
import pandas as pd

# Load the Excel file into a pandas DataFrame
file_path = '/content/drive/MyDrive/Colab Notebooks/Dataset for Data Analytics.xlsx'
df = pd.read_excel(file_path)

# Display the first 5 rows of the DataFrame
print("First 5 rows of the dataset:")
display(df.head())
```

### 2. DataFrame Information Overview

This cell provides a concise summary of the DataFrame, including the data types of each column, the number of non-null values, and memory usage. This helps us quickly identify potential data quality issues like missing values or incorrect data types.

```python
# Display basic information about the DataFrame, including data types and non-null values
print("\nDataFrame Information:")
df.info()
```

### 3. Descriptive Statistics for Numerical Columns

Here, we calculate descriptive statistics (like mean, median, standard deviation, min, max, etc.) specifically for the numerical columns in the DataFrame. This gives us insights into the central tendency, dispersion, and shape of the distribution of our quantitative data.

```python
# Display descriptive statistics for numerical columns
print("\nDescriptive statistics for numerical columns:")
display(df.describe())
```

### 4. Descriptive Statistics for All Columns

This cell expands on the previous one by providing descriptive statistics for *all* columns, including categorical ones. For categorical columns, it shows statistics like the count of unique values, the most frequent value, and its frequency. This helps in understanding the overall composition and characteristics of both numerical and non-numerical features.

```python
# Display descriptive statistics for all columns, including categorical
print("\nDescriptive statistics for all columns (including categorical):")
display(df.describe(include='all'))
```

### Summary of Patterns, Trends, and Distributions from EDA on Original Dataset
